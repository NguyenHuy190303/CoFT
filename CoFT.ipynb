```python
print("🔧 FIXING DATASET PATHS AND EXTRACTION")
print("=" * 50)

# Extract ALL datasets with proper naming convention
import tarfile
import os

# Function to extract datasets with correct naming
def extract_dataset(tar_file, extract_dir, target_name):
    """Extract dataset and ensure proper directory naming"""
    print(f"📦 Extracting {tar_file}...")
    
    if os.path.exists(tar_file):
        # Create target directory if it doesn't exist
        target_path = f"data/{target_name}"
        os.makedirs(target_path, exist_ok=True)
        
        # Extract tar file 
        with tarfile.open(tar_file, 'r:gz') as tar:
            tar.extractall(path="data/")
        
        # Check what was extracted and move if needed
        extracted_dirs = [d for d in os.listdir("data/") if os.path.isdir(f"data/{d}") and d not in [target_name, "SleepEDF"]]
        
        if extracted_dirs:
            extracted_dir = extracted_dirs[0]
            if extracted_dir != target_name:
                print(f"🔄 Moving data/{extracted_dir} → data/{target_name}")
                if os.path.exists(f"data/{extracted_dir}"):
                    # Move contents from extracted dir to target dir
                    import shutil
                    for item in os.listdir(f"data/{extracted_dir}"):
                        src = f"data/{extracted_dir}/{item}"
                        dst = f"data/{target_name}/{item}"
                        if os.path.isdir(src):
                            if os.path.exists(dst):
                                shutil.rmtree(dst)
                            shutil.move(src, dst)
                        else:
                            if os.path.exists(dst):
                                os.remove(dst)
                            shutil.move(src, dst)
                    # Remove empty extracted dir
                    os.rmdir(f"data/{extracted_dir}")
        
        print(f"✅ {target_name} extracted successfully")
        
        # Verify files
        if os.path.exists(f"data/{target_name}/train.pt"):
            print(f"   ✅ train.pt found")
        else:
            print(f"   ❌ train.pt missing")
            
    else:
        print(f"❌ {tar_file} not found")

# Extract all datasets with correct naming
extract_dataset("data/sleep.tar.gz", "data/", "sleep")
extract_dataset("data/har.tar.gz", "data/", "HAR") 
extract_dataset("data/epilepsy.tar.gz", "data/", "Epilepsy")

print("\n🔍 Final directory structure:")
!ls -la data/

print("\n🔍 Checking dataset contents:")
for dataset in ['sleep', 'HAR', 'Epilepsy']:
    dataset_path = f"data/{dataset}"
    if os.path.exists(dataset_path):
        files = os.listdir(dataset_path)
        print(f"   📂 {dataset}: {len(files)} files")
        if 'train.pt' in files:
            print(f"      ✅ train.pt available")
        else:
            print(f"      ❌ train.pt missing")
    else:
        print(f"   ❌ {dataset}: Directory not found")

print("\n✅ Dataset extraction and path fixing completed!")
```
