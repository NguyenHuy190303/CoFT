```python
# 📥 CELL 3: Download Datasets (Bypass Git LFS)
print("📥 DOWNLOADING DATASETS FROM GOOGLE DRIVE...")
print("=" * 50)

# Install gdown for Google Drive downloads
subprocess.run([sys.executable, '-m', 'pip', 'install', 'gdown'], check=True)

# Google Drive dataset links
datasets = {
    'HAR': 'https://drive.google.com/file/d/1iQTvX7CUTQrdlH5bytrGbugSbzl4x8ca/view?usp=sharing',
    'sleep': 'https://drive.google.com/file/d/1vHiLhh29q2u4WSYXM1K5JEHef8tleT2_/view?usp=sharing',
    'epilepsy': 'https://drive.google.com/file/d/1atBeEOPRTTdoIb1EwNFAjvkS8bsBwXG6/view?usp=sharing'
}

# Create data directories
os.makedirs('data/HAR', exist_ok=True)
os.makedirs('data/sleep', exist_ok=True) 
os.makedirs('data/epilepsy', exist_ok=True)

# Download and extract datasets
import tarfile

for dataset, url in datasets.items():
    print(f"📥 Downloading {dataset} dataset...")
    
    # Extract file ID from Google Drive URL
    file_id = url.split('/d/')[1].split('/')[0]
    download_url = f"https://drive.google.com/uc?id={file_id}"
    
    # Download
    tar_file = f'data/{dataset}.tar.gz'
    subprocess.run(['gdown', download_url, '-O', tar_file], check=True)
    
    # Extract
    with tarfile.open(tar_file, 'r:gz') as tar:
        tar.extractall(path=f'data/{dataset}')
    
    # Clean up
    os.remove(tar_file)
    
    # Verify
    data_dir = f'data/{dataset}'
    pt_files = [f for f in os.listdir(data_dir) if f.endswith('.pt')]
    print(f"✅ {dataset}: {len(pt_files)} .pt files extracted")

print("\n🎉 ALL DATASETS READY!")
print("📁 Sample data structure:")

# Show sample files
for dataset in ['HAR', 'sleep', 'epilepsy']:
    data_dir = f'data/{dataset}'
    if os.path.exists(data_dir):
        files = [f for f in os.listdir(data_dir) if f.endswith('.pt')][:3]
        print(f"   {dataset}: {files}")

print("\n✅ Ready for optimization!")
```

```python
# 📊 CELL 8: Analyze 3-Domain Results & Export Best Config
print("📊 ANALYZING 3-DOMAIN OPTIMIZATION RESULTS")
print("=" * 50)

import pandas as pd
import json
import glob
from datetime import datetime

# Find latest results directory
result_dirs = glob.glob("a100_results_*")
if result_dirs:
    # Sort by modification time (latest first)
    latest_dir = max(result_dirs, key=os.path.getmtime)
    print(f"📁 Latest results: {latest_dir}")
    
    # Load results CSV
    csv_file = f"{latest_dir}/results.csv"
    if os.path.exists(csv_file):
        df = pd.read_csv(csv_file)
        print(f"📋 Loaded {len(df)} experiments")
        
        # Filter successful experiments
        successful = df[df['accuracy'].notna() & (df['accuracy'] != 'None')].copy()
        successful['accuracy'] = pd.to_numeric(successful['accuracy'], errors='coerce')
        successful = successful[successful['accuracy'].notna()]
        
        print(f"✅ Successful: {len(successful)}/{len(df)} experiments")
        
        if len(successful) > 0:
            # Top 5 results
            top5 = successful.nlargest(5, 'accuracy')
            
            print("\\n🏆 TOP 5 RESULTS:")
            for idx, row in top5.iterrows():
                print(f"   {row['accuracy']:.2f}% - λ_ct={row['lambda_ct']:.5f}, λ_cs={row['lambda_cs']:.2f}, {row['ensemble']}")
            
            # Best result overall
            best = top5.iloc[0]
            print(f"\\n🥇 OVERALL BEST: {best['accuracy']:.2f}% - λ_ct={best['lambda_ct']:.5f}, λ_cs={best['lambda_cs']:.2f}, {best['ensemble']}")
            
            # Domain analysis
            print("\\n🔬 DOMAIN ANALYSIS:")
            for ensemble in ['temporal_only', 'frequency_only', 'simple_average']:
                domain_results = successful[successful['ensemble'] == ensemble]
                if len(domain_results) > 0:
                    best_domain = domain_results.nlargest(1, 'accuracy').iloc[0]
                    avg_domain = domain_results['accuracy'].mean()
                    print(f"   🎯 {ensemble:15s}: Best={best_domain['accuracy']:.2f}%, Avg={avg_domain:.2f}% ({len(domain_results)} exp)")
                else:
                    print(f"   ❌ {ensemble:15s}: No successful experiments")
            
            # Winner analysis
            domain_winners = []
            for ensemble in ['temporal_only', 'frequency_only', 'simple_average']:
                domain_results = successful[successful['ensemble'] == ensemble]
                if len(domain_results) > 0:
                    best_domain = domain_results.nlargest(1, 'accuracy').iloc[0]
                    domain_winners.append((ensemble, best_domain['accuracy']))
            
            if domain_winners:
                domain_winners.sort(key=lambda x: x[1], reverse=True)
                print(f"\\n🏆 DOMAIN WINNER: {domain_winners[0][0]} ({domain_winners[0][1]:.2f}%)")
                
                if len(domain_winners) > 1:
                    diff = domain_winners[0][1] - domain_winners[1][1]
                    print(f"📊 Lead over 2nd place: +{diff:.2f}%")
            
            # Performance comparison  
            benchmark = 76.32
            improvement = best['accuracy'] - benchmark
            print(f"\\n📈 vs Benchmark ({benchmark:.2f}%): {improvement:+.2f}%")
            
            # Export best config
            best_config = {
                'lambda_cotraining': float(best['lambda_ct']),
                'lambda_consistency': float(best['lambda_cs']),
                'ensemble_method': best['ensemble'],
                'accuracy': float(best['accuracy']),
                'timestamp': datetime.now().isoformat()
            }
            
            config_file = f"{latest_dir}/best_config.json"
            with open(config_file, 'w') as f:
                json.dump(best_config, f, indent=2)
            print(f"\\n💾 Best config saved: {config_file}")
            
        else:
            print("❌ No successful experiments")
            
else:
    print("❌ No results found - run optimization first")

print("\\n✅ 3-Domain analysis complete!")
```
